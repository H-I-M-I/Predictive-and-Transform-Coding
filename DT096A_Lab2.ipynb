```
# DT096A Lab 2 - Run this entire cell in Google Colab (CPU runtime).
# Upload clock.png, gradients.png and ManWoman.png when prompted.
from pathlib import Path
import json, csv, heapq, bisect, shutil, inspect, platform
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image
from scipy.fft import dctn, idctn

OUT = Path('DT096A_Lab2_results'); OUT.mkdir(exist_ok=True)
INPUT = Path('lab2_inputs'); INPUT.mkdir(exist_ok=True)
NAMES = ['clock.png', 'gradients.png', 'ManWoman.png']
if not all((INPUT/n).exists() for n in NAMES):
    try:
        from google.colab import files
        uploaded = files.upload()
        for name, content in uploaded.items():
            (INPUT/Path(name).name).write_bytes(content)
    except ImportError:
        # Local execution: put the images in lab2_inputs.
        pass
missing = [n for n in NAMES if not (INPUT/n).exists()]
if missing: raise FileNotFoundError('Upload these exact filenames: ' + ', '.join(missing))
plt.rcParams.update({'font.size':9, 'figure.dpi':110})
rng = np.random.default_rng(2026)
results = {}; settings = {'seed':2026,'local_entropy_window':32,'crop_size':96,
 'transform_max_dimension':512,'block_sizes':[8,16],
 'fractions':[0.05,0.1,0.25,0.5,0.75,1.0], 'KLT_training_patches':2000}

def savefig(name):
    plt.tight_layout(); plt.savefig(OUT/(name+'.png'),dpi=180,bbox_inches='tight'); plt.show(); plt.close()
def table(name, rows):
    with (OUT/(name+'.csv')).open('w',newline='') as f:
        w=csv.DictWriter(f,fieldnames=list(rows[0])); w.writeheader(); w.writerows(rows)
    print('\n'+name)
    for row in rows: print(row)
def entropy(a):
    _, counts=np.unique(a,return_counts=True); p=counts/counts.sum()
    return float(-np.sum(p*np.log2(p)))
def local_entropy(a,win=32):
    out=np.empty(a.shape,float)
    for y in range(0,a.shape[0],win):
        for x in range(0,a.shape[1],win):
            patch=a[y:y+win,x:x+win]; out[y:y+win,x:x+win]=entropy(patch)
    return out
images={}
for name in NAMES:
    im=Image.open(INPUT/name).convert('RGBA')
    bg=Image.new('RGBA',im.size,'white'); bg.alpha_composite(im)
    images[Path(name).stem]=np.asarray(bg.convert('L'))
print('Loaded original-resolution images:', {k:a.shape for k,a in images.items()})

# SECTION 3: global entropy, local entropy and equal-sized crops.
print('\nSECTION 3 - INFORMATION VARIATION')
global_rows=[]
fig, axes=plt.subplots(3,3,figsize=(12,10))
for row,(name,a) in enumerate(images.items()):
    h=entropy(a); global_rows.append({'image':name,'height':a.shape[0],'width':a.shape[1],'entropy_bits_per_pixel':h})
    axes[row,0].imshow(a,cmap='gray',vmin=0,vmax=255);axes[row,0].set_title(name);axes[row,0].axis('off')
    axes[row,1].bar(np.arange(256),np.bincount(a.ravel(),minlength=256),width=1)
    axes[row,1].set(xlabel='Gray value',ylabel='Pixel count',title=f'Global H = {h:.4f} bits/pixel')
    m=axes[row,2].imshow(local_entropy(a),cmap='magma',vmin=0,vmax=8)
    axes[row,2].set_title('Local entropy, 32 x 32 tiles');axes[row,2].axis('off');fig.colorbar(m,ax=axes[row,2],label='bits/pixel')
savefig('01_global_histograms_entropy_maps');table('global_entropy',global_rows)
# Three visually identified regions within the SAME ManWoman image.
# Coordinates are (top,left); fixed original-resolution 96 x 96 squares.
crops={'smooth background':(100,20),'edge-rich eye':(255,250),'textured hair':(40,250)}
a=images['ManWoman'];crop_rows=[]
fig,ax=plt.subplots(figsize=(9,7));ax.imshow(a,cmap='gray',vmin=0,vmax=255)
for label,(y,x) in crops.items():
    ax.add_patch(Rectangle((x,y),96,96,fill=False,edgecolor='yellow',linewidth=2));ax.text(x,y-8,label,color='yellow')
ax.axis('off');ax.set_title('Equal-sized crops in ManWoman: 96 x 96 pixels');savefig('02_crop_locations')
fig,axes=plt.subplots(3,2,figsize=(10,8))
for i,(label,(y,x)) in enumerate(crops.items()):
    c=a[y:y+96,x:x+96]; h=entropy(c)
    crop_rows.append({'region':label,'top':y,'left':x,'size':96,'entropy_bits_per_pixel':h})
    axes[i,0].imshow(c,cmap='gray',vmin=0,vmax=255);axes[i,0].set_title(label);axes[i,0].axis('off')
    axes[i,1].bar(np.arange(256),np.bincount(c.ravel(),minlength=256),width=1)
    axes[i,1].set(xlabel='Gray value',ylabel='Pixel count',title=f'H = {h:.4f} bits/pixel')
savefig('03_crop_histograms');table('crop_entropy',crop_rows)

# SECTION 4: causal predictors; first pixel raw, top row left, first column top.
# This boundary rule permits exact lossless reconstruction.
def predict(a,kind):
    a=a.astype(np.int16); p=np.zeros_like(a)
    p[0,1:]=a[0,:-1];p[1:,0]=a[:-1,0]
    L=a[1:,:-1];T=a[:-1,1:];TL=a[:-1,:-1]
    if kind=='left': p[1:,1:]=L
    elif kind=='average': p[1:,1:]=np.rint((L+T)/2).astype(np.int16)
    else: p[1:,1:]=np.where(TL>=np.maximum(L,T),np.minimum(L,T),np.where(TL<=np.minimum(L,T),np.maximum(L,T),L+T-TL))
    return p

def conditional(a):
    x=a[:,:-1].ravel().astype(np.int32);y=a[:,1:].ravel().astype(np.int32)
    joint=np.bincount(x*256+y,minlength=65536).reshape(256,256).astype(float)
    joint/=joint.sum();px=joint.sum(axis=1); mask=joint>0
    ratio=joint/np.maximum(px[:,None],1e-300)
    return float(-np.sum(joint[mask]*np.log2(ratio[mask])))

print('\nSECTION 4 - PREDICTION')
pred_rows=[];cond_rows=[];residuals={}
for name,a in images.items():
    fig,axes=plt.subplots(3,2,figsize=(11,9))
    for i,kind in enumerate(['left','average','JPEG-LS']):
        r=a.astype(np.int16)-predict(a,kind);residuals[(name,kind)]=r
        hr=entropy(r);clip=np.clip(r+128,0,255).astype(np.uint8)
        pred_rows.append({'image':name,'predictor':kind,'H_original':entropy(a),'H_signed_residual':hr,'H_clipped_display':entropy(clip),'clipped_fraction':float(np.mean((r<-128)|(r>127)))})
        lim=max(1,float(np.percentile(np.abs(r),99)))
        m=axes[i,0].imshow(r,cmap='coolwarm',vmin=-lim,vmax=lim);axes[i,0].axis('off');axes[i,0].set_title(f'{name}: {kind} residual (display saturated)');fig.colorbar(m,ax=axes[i,0],label='Signed error')
        axes[i,1].bar(np.arange(256),np.bincount(clip.ravel(),minlength=256),width=1)
        axes[i,1].set(xlabel='Residual + 128, clipped to [0,255] for display',ylabel='Count',title=f'True signed H(R) = {hr:.4f}')
    savefig('04_residuals_'+name)
    # Compare on exactly the same horizontal neighbor pairs (excludes first column).
    pair_r=a[:,1:].astype(np.int16)-a[:,:-1].astype(np.int16)
    cond_rows.append({'image':name,'H_Y_given_X':conditional(a),'H_left_residual_same_pairs':entropy(pair_r),'difference':entropy(pair_r)-conditional(a)})
table('predictor_entropy',pred_rows);table('conditional_entropy',cond_rows)

# SECTION 5: transform experiments on explicitly resized grayscale images.
# Both transforms use the same blocks. DC is the first coefficient for both.
# KLT learns zero-mean patches; the normalized constant basis restores block means.
def zigzag(N):
    pairs=sorted([(i,j) for i in range(N) for j in range(N)],key=lambda z:(sum(z),z[0] if sum(z)%2 else -z[0]))
    return np.array([i*N+j for i,j in pairs])
def blocks(a,N):
    h,w=a.shape; return a.reshape(h//N,N,w//N,N).transpose(0,2,1,3).reshape(-1,N,N)
def unblock(b,h,w,N):return b.reshape(h//N,w//N,N,N).transpose(0,2,1,3).reshape(h,w)
def klt_basis(a,N):
    patches=[]
    for _ in range(2000):
        y=rng.integers(a.shape[0]-N+1);x=rng.integers(a.shape[1]-N+1)
        p=a[y:y+N,x:x+N].ravel().astype(float);patches.append(p-p.mean())
    X=np.array(patches);cov=np.cov(X,rowvar=False)
    # Restrict covariance to the subspace orthogonal to the DC vector.
    dc=np.ones(N*N)/N
    Q=np.linalg.qr(np.column_stack([dc,np.eye(N*N)]),mode='reduced')[0][:,1:]
    vals,V=np.linalg.eigh(Q.T@cov@Q);order=np.argsort(vals)[::-1]
    basis=np.column_stack([dc,Q@V[:,order]])
    assert np.allclose(basis.T@basis,np.eye(N*N),atol=1e-9)
    return basis,vals[order]
print('\nSECTION 5 - DCT AND KLT')
transform_rows=[]
for name,original in images.items():
    scale=min(1,512/max(original.shape));h=max(16,int(original.shape[0]*scale)//16*16);w=max(16,int(original.shape[1]*scale)//16*16)
    a=np.array(Image.fromarray(original).resize((w,h),Image.Resampling.LANCZOS),dtype=float)
    Image.fromarray(a.astype(np.uint8)).save(OUT/('transform_input_'+name+'.png'))
    curves=[]
    for N in [8,16]:
        b=blocks(a,N);basis,eigenvalues=klt_basis(a,N);order=zigzag(N)
        d=dctn(b,axes=(-2,-1),norm='ortho').reshape(len(b),-1)[:,order]
        k=b.reshape(len(b),-1)@basis
        fig,axes=plt.subplots(2,8,figsize=(13,4))
        for j in range(8):
            unit=np.zeros((N,N));unit.flat[order[j]]=1;db=idctn(unit,norm='ortho')
            for row,bb in enumerate([db,basis[:,j].reshape(N,N)]):
                axes[row,j].imshow(bb,cmap='coolwarm',vmin=-np.max(abs(bb)),vmax=np.max(abs(bb)));axes[row,j].axis('off');axes[row,j].set_title(('DCT' if row==0 else 'KLT')+f' #{j}')
        fig.suptitle(f'{name}, N={N}: first basis is DC; remaining bases ordered for retention');savefig(f'05_bases_{name}_{N}')
        fig,axes=plt.subplots(2,4,figsize=(13,7))
        for kind,coef in [('DCT',d),('KLT',k)]:
            mse_list=[];energy_list=[]
            for fraction in settings['fractions']:
                keep=max(1,int(np.ceil(fraction*N*N)));actual=keep/(N*N)
                retained=np.zeros_like(coef);retained[:,:keep]=coef[:,:keep]
                if kind=='DCT':
                    normal=np.empty_like(retained);normal[:,order]=retained
                    recblocks=idctn(normal.reshape(-1,N,N),axes=(-2,-1),norm='ortho')
                else:recblocks=(retained@basis.T).reshape(-1,N,N)
                recon=unblock(recblocks,h,w,N);mse=float(np.mean((a-recon)**2));den=float(np.sum(coef**2));energy=float(np.sum(retained**2)/den) if den else 1.0
                acden=float(np.sum(coef[:,1:]**2));acenergy=float(np.sum(retained[:,1:]**2)/acden) if acden else 1.0
                if fraction==1: assert mse<1e-16, (kind,mse)
                transform_rows.append({'image':name,'height':h,'width':w,'N':N,'transform':kind,'requested_fraction':fraction,'coefficients_kept':keep,'actual_fraction':actual,'MSE':mse,'retained_total_energy_fraction':energy,'retained_AC_energy_fraction':acenergy})
                mse_list.append(mse);energy_list.append(acenergy)
                if fraction in [0.05,0.1,0.25,1.0]:
                    col=[0.05,0.1,0.25,1.0].index(fraction);row=0 if kind=='DCT' else 1
                    axes[row,col].imshow(recon,cmap='gray',vmin=0,vmax=255);axes[row,col].axis('off');axes[row,col].set_title(f'{kind}, {keep}/{N*N}\nMSE={mse:.2f}')
            curves.append((N,kind,mse_list,energy_list))
        fig.suptitle(f'{name}: reconstructed images, block size {N}');savefig(f'06_reconstructions_{name}_{N}')
    fig,axes=plt.subplots(1,2,figsize=(11,4))
    for N,kind,ms,es in curves:
        fs=[np.ceil(f*N*N)/(N*N) for f in settings['fractions']]
        axes[0].plot(fs,ms,'o-',label=f'{kind}, {N}x{N}');axes[1].plot(fs,es,'o-',label=f'{kind}, {N}x{N}')
    axes[0].set(xlabel='Actual retained coefficient fraction',ylabel='MSE (gray-value squared)',title=name+' reconstruction error')
    axes[1].set(xlabel='Actual retained coefficient fraction',ylabel='Retained AC energy fraction',title='Energy compaction excluding DC')
    for ax in axes:ax.legend();ax.grid(alpha=.3)
    savefig('07_transform_curves_'+name)
table('transform_results',transform_rows)

# SECTION 6: real static Huffman and finite-precision arithmetic bitstreams.
# Both use the identical empirical frequency model. Length excludes model/header.
def huffman(counts):
    heap=[(int(c),i,i) for i,c in enumerate(counts) if c];heapq.heapify(heap);serial=len(counts)
    while len(heap)>1:
        ca,_,a=heapq.heappop(heap);cb,_,b=heapq.heappop(heap)
        heapq.heappush(heap,(ca+cb,serial,(a,b)));serial+=1
    codes={}
    def walk(node,s):
        if isinstance(node,int):codes[node]=s or '0'
        else:walk(node[0],s+'0');walk(node[1],s+'1')
    walk(heap[0][2],'');return codes

def arithmetic_encode(data,counts):
    cum=np.r_[0,np.cumsum(counts)].tolist();total=cum[-1]
    FULL=1<<32;HALF=FULL>>1;Q=HALF>>1;low=0;high=FULL-1;pending=0;bits=[]
    def emit(bit):
        nonlocal pending
        bits.append(bit);bits.extend([1-bit]*pending);pending=0
    for s in data:
        s=int(s);span=high-low+1
        high=low+span*cum[s+1]//total-1;low=low+span*cum[s]//total
        while True:
            if high<HALF:emit(0)
            elif low>=HALF:emit(1);low-=HALF;high-=HALF
            elif low>=Q and high<3*Q:pending+=1;low-=Q;high-=Q
            else:break
            low*=2;high=high*2+1
    pending+=1;emit(0 if low<Q else 1)
    return bits

def arithmetic_decode(bits,counts,n):
    cum=np.r_[0,np.cumsum(counts)].tolist();total=cum[-1]
    FULL=1<<32;HALF=FULL>>1;Q=HALF>>1;low=0;high=FULL-1;pos=0
    def read():
        nonlocal pos
        bit=bits[pos] if pos<len(bits) else 0;pos+=1;return bit
    value=0
    for _ in range(32):value=value*2+read()
    decoded=[]
    for _ in range(n):
        span=high-low+1;scaled=((value-low+1)*total-1)//span
        s=bisect.bisect_right(cum,scaled)-1;decoded.append(s)
        high=low+span*cum[s+1]//total-1;low=low+span*cum[s]//total
        while True:
            if high<HALF:pass
            elif low>=HALF:low-=HALF;high-=HALF;value-=HALF
            elif low>=Q and high<3*Q:low-=Q;high-=Q;value-=Q
            else:break
            low*=2;high=high*2+1;value=value*2+read()
    return np.array(decoded)
def packed(bits,path):
    (OUT/path).write_bytes(np.packbits(np.asarray(bits,dtype=np.uint8)).tobytes())

print('\nSECTION 6 - ENTROPY CODING')
coding_rows=[]
for name,a in images.items():
    # Fixed central 256x256 image crop: prediction performed on this standalone crop.
    y=(a.shape[0]-256)//2;x=(a.shape[1]-256)//2;c=a[y:y+256,x:x+256]
    r=c.astype(np.int16)-predict(c,'JPEG-LS');symbols,stream=np.unique(r.ravel(),return_inverse=True)
    counts=np.bincount(stream);p=counts/counts.sum();H=entropy(stream)
    codes=huffman(counts);hbits=''.join(codes[int(s)] for s in stream)
    inverse={v:k for k,v in codes.items()};decoded=[];prefix=''
    for bit in hbits:
        prefix+=bit
        if prefix in inverse:decoded.append(inverse[prefix]);prefix=''
    assert not prefix and np.array_equal(decoded,stream)
    abits=arithmetic_encode(stream,counts);assert np.array_equal(arithmetic_decode(abits,counts,len(stream)),stream)
    packed([int(b) for b in hbits],name+'_huffman.bin');packed(abits,name+'_arithmetic.bin')
    model={'symbols':symbols.tolist(),'counts':counts.tolist(),'huffman_codes':codes,'n_symbols':len(stream),'huffman_valid_bits':len(hbits),'arithmetic_valid_bits':len(abits),'crop_top':y,'crop_left':x,'crop_size':256}
    (OUT/(name+'_coding_model.json')).write_text(json.dumps(model,indent=2))
    table(name+'_source_pmf',[{'signed_residual':int(s),'count':int(c),'probability':float(pp),'huffman_code':codes[i]} for i,(s,c,pp) in enumerate(zip(symbols,counts,p))])
    coding_rows.append({'image':name,'source':'JPEG-LS residuals, central 256x256 crop','symbols':len(stream),'alphabet_size':len(counts),'entropy_bits_per_symbol':H,'Huffman_bits_per_symbol':len(hbits)/len(stream),'arithmetic_bits_per_symbol':len(abits)/len(stream),'Huffman_bits':len(hbits),'arithmetic_bits':len(abits),'both_decode_exactly':True,'model_header_included':False})
    fig,axes=plt.subplots(1,2,figsize=(11,4));axes[0].bar(symbols,p);axes[0].set(xlabel='Signed residual value',ylabel='Probability',title=name+' source PMF')
    axes[1].bar(['Entropy','Huffman','Arithmetic'],[H,len(hbits)/len(stream),len(abits)/len(stream)]);axes[1].set(ylabel='bits/symbol',title='Payload rate (model/header excluded)');savefig('08_coding_'+name)
table('coding_comparison',coding_rows)

# Save methods, report snippets, checks, numerical tables and all figures.
results={'settings':settings,'global_entropy':global_rows,'crop_entropy':crop_rows,'prediction':pred_rows,'conditional_entropy':cond_rows,'transforms':transform_rows,'coding':coding_rows}
(OUT/'all_results.json').write_text(json.dumps(results,indent=2))
snippets={'3_information':entropy,'4_prediction':predict,'5_transform':klt_basis,'6_entropy_coding':huffman}
for name,fn in snippets.items():
    try:text=inspect.getsource(fn)
    except (OSError,TypeError):text='See corresponding function in the notebook: '+fn.__name__
    (OUT/(name+'_snippet.py')).write_text(text)
notes='''METHODS AND REPORT CHECKLIST
Report maximum: 5 pages, including code snippets (no appendix exemption stated).
Use key labeled figures, 2-5-sentence result interpretations, 1-2 code snippets
per section, and a 5-8-sentence reflection connecting experiments to lectures.
External support: code prepared with ChatGPT/Codex assistance; lab PDF starter
concepts; NumPy for arrays and covariance, Pillow for grayscale and resizing,
SciPy for orthonormal DCT/IDCT, Matplotlib for figures; Python standard library
for Huffman trees and file operations. No third-party entropy coder was used.
Global and prediction experiments use original-resolution grayscale inputs.
Local entropy uses tiled windows, including smaller tiles at image boundaries.
Crop labels identify visual structures; observed entropy need not follow label order.
Predictor boundary handling is causal: first pixel predicted as 0, first row from
left, first column from top. Arithmetic and Huffman decoding have been verified.
Residual histogram displayed after +128/clipping as requested, but actual entropy
and coding use signed residuals without clipping. Both values are recorded.
Conditional entropy and comparison residual entropy use identical horizontal pairs.
Transforms use resized images of maximum dimension 512, sizes divisible by 16.
DCT keeps zigzag coefficients; KLT keeps DC plus eigenvectors by descending variance.
KLT training patches have their own mean removed; block means are retained as DC.
KLT is trained on the evaluated image (in-image, not independent-test performance).
KLT basis storage is not included: this is a transform comparison, not a codec rate.
Coefficient count includes DC, with ceil rounding. MSE uses floating reconstructions
before display clipping; full-coefficient reconstruction is checked for exactness.
Both total and AC-only retained energy are recorded (DC can dominate total energy).
Entropy coding uses standalone central 256x256 crops with JPEG-LS residuals.
Both coders use the same static empirical counts. Actual valid bit lengths include
arithmetic termination but exclude byte padding, frequency-model and header costs.
Binary streams, valid bit counts and models are saved for reproducibility.
Interpret these as zero-order symbol entropy rates, not full-image compression ratios.
'''
(OUT/'METHODS_AND_REPORT_CHECKLIST.txt').write_text(notes+'\nPython '+platform.python_version()+'; NumPy '+np.__version__)
archive=shutil.make_archive(str(OUT),'zip',OUT)
print('\nCOMPLETE. Download DT096A_Lab2_results.zip and upload it to ChatGPT for the report.')
try:
    from google.colab import files
    files.download(archive)
except ImportError:print('Saved:',archive)
```
